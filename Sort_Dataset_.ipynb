```python
import pandas as pd

# Create Dataset
data = {
    "Employee": ["Arun", "Priya", "Kavin", "Divya", "Meena", "Ravi"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Age": [28, 35, 24, 30, 27, 32],
    "Salary": [45000, 60000, 38000, 52000, 48000, 55000]
}

# Convert data into DataFrame
df = pd.DataFrame(data)

# Display Original Dataset
print("Original Dataset")
print(df)

# 1. Ascending Order
print("\n1. Salary - Ascending Order")
print(df.sort_values(by="Salary", ascending=True))

# 2. Descending Order
print("\n2. Salary - Descending Order")
print(df.sort_values(by="Salary", ascending=False))

# 3. Sort by Department
print("\n3. Department - Alphabetical Order")
print(df.sort_values(by="Department", ascending=True))

# 4. Sort by Multiple Columns
print("\n4. Sort by Department and Salary")
print(
    df.sort_values(
        by=["Department", "Salary"],
        ascending=[True, False]
    )
)

# 5. Sort by Age
print("\n5. Age - Ascending Order")
print(df.sort_values(by="Age", ascending=True))
```
